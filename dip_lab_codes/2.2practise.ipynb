```python
import math
import numpy as np
import PIL.Image as Image
import matplotlib.pyplot as plt
```

```python
img = Image.open("images/vu.jpg").convert("L")
img = np.array(img)

H, W = img.shape
plt.title("Grayschal Image")
plt.imshow(img, cmap="gray")
plt.show()
```

```python
negative_img = np.zeros(shape=img.shape)

for y in range(H):
    for x in range(W):
        negative_img[y,x]=255- img[y,x]

fig,axes = plt.subplots(ncols=2,nrows=1,figsize=(15,10))

axes[0].set_title("Grayschale Image")
axes[0].imshow(img, cmap="gray")
axes[1].set_title("Negative Image")
axes[1].imshow(negative_img, cmap="gray")

plt.show()

```

```python
log_image = np.zeros(shape=img.shape)

c = 255/math.log(1+255)

for y in range(H):
    for x in range(W):
        if img[y,x]!=255:
            log_image[y,x] = c *math.log(1+img[y,x]) 
        else:
            log_image[y,x] = c*math.log(img[y,x])


fig,axes = plt.subplots(ncols=2,nrows=1,figsize=(15,10))

axes[0].set_title("Grayschale Image")
axes[0].imshow(img, cmap="gray")

axes[1].set_title("Grayschale Image")
axes[1].imshow(log_image, cmap="gray")

plt.show()
```

```python
gamma_img = np.zeros(shape=img.shape)
gamma=2
c=1/255

for y in range(H):
    for x in range(W):
        gamma_img[y,x] = c * (img[y,x]**gamma)

fig,axes = plt.subplots(ncols=2, nrows=1, figsize=(15,10))

axes[0].set_title("Grayschale Image")
axes[0].imshow(img, cmap="gray")



axes[1].set_title("gamma image")
axes[1].imshow(gamma_img, cmap="gray")
plt.show()

```

```python
binary_img = np.zeros(shape=img.shape)

threshold=120

for y in range(H):
    for x in range(W):
        if img[y,x]>=threshold:
            binary_img[y,x]=255
        else: 
            0

fig,axes=plt.subplots(ncols=2,nrows=1,figsize=(15,10))

axes[0].set_title("Grayschale Image")
axes[0].imshow(img, cmap="gray")



axes[1].set_title("gamma image")
axes[1].imshow(binary_img, cmap="gray")
plt.show()

```
