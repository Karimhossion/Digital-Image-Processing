```python
import math
import numpy as np
import PIL.Image as Image
import matplotlib.pyplot as plt
```

```python
img = Image.open("images/vu.jpg").convert("L")
img = np.array(img)

H, W = img.shape

plt.title("Grayscale")
plt.imshow(img, cmap="gray")
plt.show()
```

```python
F=np.fft.fft2(img)
F_shift=np.fft.fftshift(F)

real = F_shift.real
imag = F_shift.imag

magnitude=(real**2+imag**2)**0.5

log=(1/math.log(1+255))*np.log(1+magnitude)
fig,axes=plt.subplots(ncols=2,nrows=1,figsize=(15,10))

axes[0].set_title("Grayscale")
axes[0].imshow(img, cmap="gray")

axes[1].set_title("Frequency Specturm")
axes[1].imshow(log, cmap="gray")

plt.show()
```
