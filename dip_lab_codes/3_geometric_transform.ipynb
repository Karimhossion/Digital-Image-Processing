```python
import math
import numpy as np
import PIL.Image as Image
import matplotlib.pyplot as plt
```

```python
img = Image.open("images/vu.jpg").convert("L")
img = np.array(img)

H, W = img.shape

plt.title("Grayscale")
plt.imshow(img, cmap="gray")
plt.show()
```

```python
scaled_img = np.zeros(shape=img.shape)
sx = 1
sy = 2
scaling_matrix = np.array(
    [
        [sy, 0],
        [0, sx]
    ]
)

for y in range(H):
    for x in range(W):
        old_idx = np.array([y, x])
        y_new, x_new = np.matmul(scaling_matrix, old_idx)
        y_new, x_new = int(y_new), int(x_new)

        if (y_new >= 0 and y_new < H) and (x_new >= 0 and x_new < W):
            scaled_img[y_new, x_new] = img[y, x]

fig, axes = plt.subplots(ncols=2, nrows=1, figsize=(15, 10))

axes[0].set_title("Grayscale")
axes[0].imshow(img, cmap="gray")

axes[1].set_title("Scaled")
axes[1].imshow(scaled_img, cmap="gray")

plt.show()
```

```python
rotated_img = np.zeros(shape=img.shape)
angle = 10
theta = (math.pi * angle) / 180

rotation_matrix = np.array(
    [
        [math.cos(theta), -math.sin(theta)],
        [math.sin(theta), math.cos(theta)]
    ]
)

for y in range(H):
    for x in range(W):
        old_idx = np.array([y, x])
        y_new, x_new = np.matmul(rotation_matrix, old_idx)
        y_new, x_new = int(y_new), int(x_new)

        if (y_new >= 0 and y_new < H) and (x_new >= 0 and x_new < W):
            rotated_img[y_new, x_new] = img[y, x]

fig, axes = plt.subplots(ncols=2, nrows=1, figsize=(15, 10))

axes[0].set_title("Grayscale")
axes[0].imshow(img, cmap="gray")

axes[1].set_title("Rotated")
axes[1].imshow(rotated_img, cmap="gray")

plt.show()
```

```python
sheared_img = np.zeros(shape=img.shape)
shx = 0.5
shy = 0.8
shearing_matrix = np.array(
    [
        [1, shy],
        [shx, 1]
    ]
)

for y in range(H):
    for x in range(W):
        old_idx = np.array([y, x])
        y_new, x_new = np.matmul(shearing_matrix, old_idx)
        y_new, x_new = int(y_new), int(x_new)

        if (y_new >= 0 and y_new < H) and (x_new >= 0 and x_new < W):
            sheared_img[y_new, x_new] = img[y, x]

fig, axes = plt.subplots(ncols=2, nrows=1, figsize=(15, 10))

axes[0].set_title("Grayscale")
axes[0].imshow(img, cmap="gray")

axes[1].set_title("Sheared")
axes[1].imshow(sheared_img, cmap="gray")

plt.show()
```

```python
translated_img = np.zeros(shape=img.shape)
tx = 200
ty = 150
translation_matrix = np.array(
    [ty, tx]
)

for y in range(H):
    for x in range(W):
        old_idx = np.array([y, x])
        y_new, x_new = translation_matrix + old_idx
        y_new, x_new = int(y_new), int(x_new)

        if (y_new >= 0 and y_new < H) and (x_new >= 0 and x_new < W):
            translated_img[y_new, x_new] = img[y, x]

fig, axes = plt.subplots(ncols=2, nrows=1, figsize=(15, 10))

axes[0].set_title("Grayscale")
axes[0].imshow(img, cmap="gray")

axes[1].set_title("Translated")
axes[1].imshow(translated_img, cmap="gray")

plt.show()
```
