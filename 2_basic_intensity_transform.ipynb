```python
import math
import numpy as np
import PIL.Image as Image
import matplotlib.pyplot as plt
```

```python
img = Image.open("images/vu.jpg").convert("L")
img = np.array(img)

H, W = img.shape

plt.title("Grayscale")
plt.imshow(img, cmap="gray")
plt.show()
```

```python
negative_img = np.zeros(shape=img.shape)

for y in range(H):
    for x in range(W):
        negative_img[y, x] = 255 - img[y, x]

fig, axes = plt.subplots(ncols=2, nrows=1, figsize=(15, 10))

axes[0].set_title("Grayscale")
axes[0].imshow(img, cmap="gray")

axes[1].set_title("Negative")
axes[1].imshow(negative_img, cmap="gray")

plt.show()
```

```python
log_img = np.zeros(shape=img.shape)

c = 255 / math.log(1 + 255)

for y in range(H):
    for x in range(W):
        log_img[y, x] = c * math.log(1 + img[y, x]) if img[y, x] != 255 else c * math.log(img[y, x])

fig, axes = plt.subplots(ncols=2, nrows=1, figsize=(15, 10))

axes[0].set_title("Grayscale")
axes[0].imshow(img, cmap="gray")

axes[1].set_title("Log")
axes[1].imshow(log_img, cmap="gray")

plt.show()
```

```python
gamma_img = np.zeros(shape=img.shape)
gamma = 2
c = 1 / 255

for y in range(H):
    for x in range(W):
        gamma_img[y, x] = c * ( img[y, x] ** gamma)

fig, axes = plt.subplots(ncols=2, nrows=1, figsize=(15, 10))

axes[0].set_title("Grayscale")
axes[0].imshow(img, cmap="gray")

axes[1].set_title("Gamma")
axes[1].imshow(gamma_img, cmap="gray")

plt.show()
```

```python
bin_img = np.zeros(shape=img.shape)
threshold = 128

for y in range(H):
    for x in range(W):
        bin_img[y, x] = 255 if img[y, x] >= threshold else 0

fig, axes = plt.subplots(ncols=2, nrows=1, figsize=(15, 10))

axes[0].set_title("Grayscale")
axes[0].imshow(img, cmap="gray")

axes[1].set_title("Binary")
axes[1].imshow(bin_img, cmap="gray")

plt.show()
```
